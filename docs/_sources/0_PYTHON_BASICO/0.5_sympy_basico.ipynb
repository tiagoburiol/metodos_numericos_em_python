# 0.5 SymPy Básico

**SymPy** é uma biblioteca Python para matemática simbólica. Ao contrário do NumPy, que trabalha com valores numéricos aproximados, o SymPy manipula expressões matemáticas de forma exata — como um sistema de álgebra computacional (CAS).

Isso é útil para:
- Simplificar e expandir expressões algébricas
- Calcular derivadas e integrais analiticamente
- Resolver equações de forma exata
- Calcular limites e séries de Taylor
- Verificar resultados de métodos numéricos

Neste notebook veremos os principais recursos do SymPy com exemplos práticos.

```python
import sympy as sp

# Ativa impressão formatada (LaTeX no Jupyter)
sp.init_printing(use_unicode=True)
```

## Símbolos

Em SymPy, as variáveis matemáticas são chamadas de **símbolos** e devem ser declaradas explicitamente com `sp.symbols()`.

```python
x, y, z = sp.symbols('x y z')
a, b, c = sp.symbols('a b c')

# Uma expressão simbólica
expr = x**2 + 2*x + 1
expr
```

```python
# Símbolos com restrições (positivo, inteiro, etc.)
n = sp.symbols('n', integer=True, positive=True)
r = sp.symbols('r', positive=True)
print(type(expr))
```

## Constantes simbólicas

O SymPy possui suas próprias constantes exatas — não use `math.pi` ou `math.e` em expressões simbólicas.

```python
print(sp.pi)          # π exato
print(sp.E)           # e exato
print(sp.sqrt(2))     # √2 exato
print(sp.Rational(1, 3))  # fração exata 1/3
```

```python
# Comparação: simbólico vs numérico
print('Simbólico:', sp.sqrt(2)**2)      # exato: 2
print('Numérico: ', 2**0.5 ** 2)        # ponto flutuante
```

## Simplificação e manipulação de expressões

### `expand` — expandir

Distribui produtos e potências.

```python
expr = (x + 1)**3
sp.expand(expr)
```

```python
sp.expand((x + y)**2)
```

### `factor` — fatorar

Operação inversa de `expand`: decompõe em fatores.

```python
expr = x**3 - x**2 - x + 1
sp.factor(expr)
```

```python
sp.factor(x**2 - 5*x + 6)
```

### `simplify` — simplificar

Tenta encontrar a forma mais simples de uma expressão (heurística geral).

```python
expr = (x**2 - 1) / (x - 1)
sp.simplify(expr)
```

```python
sp.simplify(sp.sin(x)**2 + sp.cos(x)**2)
```

### `subs` — substituição

Substitui um símbolo por um valor numérico ou por outra expressão.

```python
expr = x**2 + 2*x + 1

# Substituir x por 3
print(expr.subs(x, 3))

# Substituir x por y + 1
print(expr.subs(x, y + 1))
```

```python
# Avaliação numérica com evalf()
expr = sp.sqrt(2) + sp.pi
print(expr)
print(expr.evalf())          # padrão: 15 dígitos
print(expr.evalf(50))        # 50 dígitos
```

## Derivadas

Use `sp.diff(f, x)` para calcular $\frac{df}{dx}$. Para derivadas de ordem superior, informe a ordem como terceiro argumento.

```python
f = x**4 - 3*x**3 + 2*x - 5

print('f     =', f)
print("f'    =", sp.diff(f, x))
print("f''   =", sp.diff(f, x, 2))
print("f'''  =", sp.diff(f, x, 3))
```

```python
# Derivadas de funções trigonométricas e exponenciais
print(sp.diff(sp.sin(x), x))
print(sp.diff(sp.cos(x), x))
print(sp.diff(sp.exp(x), x))
print(sp.diff(sp.log(x), x))
```

```python
# Derivada parcial
g = x**2 * y + sp.sin(x * y)
print('∂g/∂x =', sp.diff(g, x))
print('∂g/∂y =', sp.diff(g, y))
```

## Integrais

Use `sp.integrate(f, x)` para integral indefinida e `sp.integrate(f, (x, a, b))` para integral definida.

```python
# Integral indefinida
f = x**3 - 2*x + 1
sp.integrate(f, x)
```

```python
# Funções elementares
print(sp.integrate(sp.sin(x), x))
print(sp.integrate(sp.exp(-x**2), x))   # função de Gauss — sem forma fechada elementar
```

```python
# Integral definida
resultado = sp.integrate(x**2, (x, 0, 3))
print('∫₀³ x² dx =', resultado)
```

```python
# Integral com limites simbólicos
a, b = sp.symbols('a b')
sp.integrate(sp.exp(-a * x), (x, 0, sp.oo))
```

## Limites

Use `sp.limit(f, x, x0)` para calcular $\lim_{x \to x_0} f(x)$.

```python
# Limite clássico: sin(x)/x quando x -> 0
sp.limit(sp.sin(x) / x, x, 0)
```

```python
# Limite no infinito
sp.limit((1 + 1/x)**x, x, sp.oo)
```

```python
# Limites laterais: '+' (direita) ou '-' (esquerda)
print(sp.limit(1/x, x, 0, '+'))
print(sp.limit(1/x, x, 0, '-'))
```

## Séries de Taylor

`f.series(x, x0, n)` calcula a expansão em série de Taylor de $f$ em torno de $x_0$ até ordem $n-1$.

```python
# Série de Taylor de e^x em torno de x=0 até ordem 6
sp.exp(x).series(x, 0, 6)
```

```python
# sin(x) em torno de x=0
sp.sin(x).series(x, 0, 8)
```

```python
# Remover o termo O (ordem de truncamento) com removeO()
sp.cos(x).series(x, 0, 7).removeO()
```

## Resolução de equações

Use `sp.solve(equacao, variavel)` para resolver equações algebricas. A equação é passada como expressão igual a zero.

```python
# Equação quadrática: x² - 5x + 6 = 0
sp.solve(x**2 - 5*x + 6, x)
```

```python
# Fórmula quadrática geral: ax² + bx + c = 0
a, b, c = sp.symbols('a b c')
sp.solve(a*x**2 + b*x + c, x)
```

```python
# Sistema de equações
x, y = sp.symbols('x y')
sol = sp.solve([x + y - 5,
                2*x - y - 1], [x, y])
print(sol)
```

## Convertendo para função numérica: `lambdify`

Uma vez que tenhamos uma expressão simbólica, é comum querer avaliá-la numericamente em muitos pontos. A função `sp.lambdify()` converte uma expressão SymPy em uma função NumPy eficiente.

```python
import numpy as np
import matplotlib.pyplot as plt
plt.rcParams['figure.figsize'] = [6, 4]

x = sp.symbols('x')
expr = sp.sin(x) * sp.exp(-x/5)

# Converter para função numérica
f = sp.lambdify(x, expr, 'numpy')

# Avaliar em um array NumPy
t = np.linspace(0, 20, 300)
plt.plot(t, f(t))
plt.title(r'$f(x) = \sin(x)\,e^{-x/5}$')
plt.xlabel('x')
plt.grid(True)
plt.show()
```

## Resumo das funções

| Operação | Função SymPy |
|---|---|
| Declarar símbolos | `sp.symbols('x y')` |
| Expandir | `sp.expand(expr)` |
| Fatorar | `sp.factor(expr)` |
| Simplificar | `sp.simplify(expr)` |
| Substituir | `expr.subs(x, valor)` |
| Avaliar numericamente | `expr.evalf(n)` |
| Derivada | `sp.diff(f, x, n)` |
| Integral indefinida | `sp.integrate(f, x)` |
| Integral definida | `sp.integrate(f, (x, a, b))` |
| Limite | `sp.limit(f, x, x0)` |
| Série de Taylor | `f.series(x, x0, n)` |
| Resolver equação | `sp.solve(eq, x)` |
| Converter para NumPy | `sp.lambdify(x, expr, 'numpy')` |
