```python
import numpy as np
import matplotlib.pyplot as plt

# Parâmetros de simulação
fs = 1000  # taxa de amostragem (Hz)
T = 2      # duração do sinal (s)
t = np.linspace(0, T, int(fs*T), endpoint=False)

# Sinal composto (somando 3 senóides)
freqs = [1, 10, 20]  # Hz
signal = sum(np.sin(2*np.pi*f*t) for f in freqs)

# FFT
N = len(t)
fft_vals = np.fft.fft(signal)
fft_freqs = np.fft.fftfreq(N, 1/fs)

# Apenas parte positiva
pos_mask = fft_freqs >= 0
fft_freqs = fft_freqs[pos_mask]
fft_vals = np.abs(fft_vals[pos_mask]) / N

# Plot do sinal no tempo
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.plot(t, signal)
plt.title("Sinal Composto (1, 10 e 20 Hz)")
plt.xlabel("Tempo [s]")
plt.ylabel("Amplitude")

# Plot do espectro (linhas espectrais)
plt.subplot(1,2,2)
plt.stem(fft_freqs, fft_vals, basefmt=" ")
plt.xlim(0, 30)
plt.title("Espectro do Sinal (Linhas Espectrais)")
plt.xlabel("Frequência [Hz]")
plt.ylabel("Amplitude")

# Marcar frequências detectadas
for f in freqs:
    idx = np.argmin(np.abs(fft_freqs - f))
    plt.annotate(f"{f} Hz", (fft_freqs[idx], fft_vals[idx]),
                 textcoords="offset points", xytext=(0,10), ha='center', color='red')

plt.tight_layout()
plt.show()
```
