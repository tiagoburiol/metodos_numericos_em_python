```python
import numpy as np
import matplotlib.pyplot as plt

# --- 1. CONFIGURAÇÕES DO MODELO DE TRÁFEGO (Seção 4.1) ---
N_ESPECIES = 2  # No artigo pode ser até 10+
V_MAX = np.array([1.0, 0.6]) # Velocidades máximas (v1 > v2)
L = 1.0 # Comprimento da estrada

def fluxo_trafego(phi):
    """ Fluxo para o modelo multi-classe LWR: fi = phi_i * v_i(rho) """
    rho = np.sum(phi, axis=0) # Densidade total
    # Velocidade comum baseada na densidade total (V = Vmax * (1 - rho))
    v = np.zeros_like(phi)
    for i in range(N_ESPECIES):
        v[i] = V_MAX[i] * (1.0 - rho)
    return phi * v

# --- 2. MOTOR WENO DE 5ª ORDEM ---
def weno5_reconstruction(f):
    """ Reconstrução WENO-JS de 5ª ordem para os fluxos """
    eps = 1e-6
    # Stencils
    p0 = (2/6)*f[0: -4] - (7/6)*f[1: -3] + (11/6)*f[2: -2]
    p1 = -(1/6)*f[1: -3] + (5/6)*f[2: -2] + (2/6)*f[3: -1]
    p2 = (2/6)*f[2: -2] + (5/6)*f[3: -1] - (1/6)*f[4:]
    
    # Indicadores de suavidade (Beta)
    b0 = (13/12)*(f[0:-4] - 2*f[1:-3] + f[2:-2])**2 + (1/4)*(f[0:-4] - 4*f[1:-3] + 3*f[2:-2])**2
    b1 = (13/12)*(f[1:-3] - 2*f[2:-2] + f[3:-1])**2 + (1/4)*(f[1:-3] - f[3:-1])**2
    b2 = (13/12)*(f[2:-2] - 2*f[3:-1] + f[4:])**2 + (1/4)*(3*f[2:-2] - 4*f[3:-1] + f[4:])**2
    
    # Pesos
    d = np.array([0.1, 0.6, 0.3])
    alpha0 = d[0] / (b0 + eps)**2
    alpha1 = d[1] / (b1 + eps)**2
    alpha2 = d[2] / (b2 + eps)**2
    
    alphasum = alpha0 + alpha1 + alpha2
    w0, w1, w2 = alpha0/alphasum, alpha1/alphasum, alpha2/alphasum
    
    return w0*p0 + w1*p1 + w2*p2

# --- 3. ALGORITMO DE MULTIRRESOLUÇÃO (2.1 & 2.2) ---
def obter_detalhes(u, Lc):
    """ Calcula os coeficientes de wavelet (detalhes) """
    detalhes = []
    u_level = np.copy(u)
    for k in range(Lc):
        u_grossa = u_level[:, ::2]
        # Interpolação de 4ª ordem (simplificada p/ 2ª nas bordas)
        u_predito = 0.5 * (u_grossa[:, :-1] + u_grossa[:, 1:])
        d = u_level[:, 1::2] - u_predito
        detalhes.append(d)
        u_level = u_grossa
    return detalhes

def gerar_malha_adaptativa(u, Lc, epsilon):
    """ Algoritmo 2.1: Marca pontos importantes """
    detalhes = obter_detalhes(u, Lc)
    mascara = np.zeros(u.shape[1], dtype=bool)
    
    for k, d in enumerate(detalhes):
        passo = 2**(k+1)
        # Se algum detalhe de qualquer espécie for > epsilon
        for j in range(d.shape[1]):
            if np.any(np.abs(d[:, j]) > epsilon):
                idx = j * passo + 1
                # Zona de segurança (vizinhos)
                mascara[max(0, idx-4):min(len(mascara), idx+5)] = True
    return mascara

# --- 4. CICLO DE EVOLUÇÃO (ALGORITMO 2.2) ---
def resolver():
    Nx = 256  # Malha fina G0
    Lc = 3    # Níveis de multirresolução
    dx = L / Nx
    dt = 0.1 * dx # Condição CFL
    epsilon = 0.005 # Tolerância de compressão
    
    x = np.linspace(0, L, Nx)
    # Condição Inicial: Um pelotão de carros no meio
    phi = np.zeros((N_ESPECIES, Nx))
    phi[0, (x > 0.2) & (x < 0.4)] = 0.4 # Espécie lenta
    phi[1, (x > 0.2) & (x < 0.3)] = 0.3 # Espécie rápida
    
    plt.ion()
    fig, ax = plt.subplots(2, 1, figsize=(10, 8))

    for t in range(150):
        # 1. Identificar malha ativa (Algoritmo 2.1)
        mask = gerar_malha_adaptativa(phi, Lc, epsilon)
        
        # 2. Integração Temporal (RK3)
        # Para simplificar, calculamos o fluxo global e aplicamos a máscara
        def compute_rhs(p):
            f = fluxo_trafego(p)
            rhs = np.zeros_like(p)
            for i in range(N_ESPECIES):
                # WENO necessita de preenchimento (ghost cells)
                f_ext = np.pad(f[i], 3, mode='wrap')
                f_reconst = weno5_reconstruction(f_ext)
                rhs[i, 1:-1] = -(f_reconst[1:] - f_reconst[:-1]) / dx
            return rhs * mask # A mágica adaptativa: só evolui onde a máscara permite

        # RK3 Stage 1
        u1 = phi + dt * compute_rhs(phi)
        # Stage 2
        u2 = 0.75*phi + 0.25*u1 + 0.25*dt*compute_rhs(u1)
        # Stage 3
        phi = (1/3)*phi + (2/3)*u2 + (2/3)*dt*compute_rhs(u2)

        # Visualização
        if t % 10 == 0:
            ax[0].cla()
            ax[0].plot(x, phi[0], label='Espécie 1 (Lenta)')
            ax[0].plot(x, phi[1], label='Espécie 2 (Rápida)')
            ax[0].set_title(f"Evolução do Fluxo - Tempo: {t*dt:.3f}")
            ax[0].legend()
            
            ax[1].cla()
            ax[1].scatter(x, np.zeros_like(x), c=mask, cmap='binary', s=1)
            ax[1].set_title("Malha Ativa (Preto = Cálculo WENO, Branco = Ignorado)")
            plt.pause(0.01)

    plt.ioff()
    plt.show()

if __name__ == "__main__":
    resolver()
```
