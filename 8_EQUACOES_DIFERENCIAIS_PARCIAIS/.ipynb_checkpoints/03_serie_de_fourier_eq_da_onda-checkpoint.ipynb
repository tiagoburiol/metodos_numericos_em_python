### Série de Fourier 

Série de Fourier da função $f(x)$ de período $2 L$

$$
f(x)=a_0+\sum_{n=1}^{\infty}\left(a_n \cos \frac{n \pi}{L} x+b_n \operatorname{sen} \frac{n \pi}{L} x\right)
$$

com os coeficientes de Fourier de $f(x)$ dados pelas fórmulas de Euler

$$
\begin{gathered}
a_0=\frac{1}{2 L} \int_{-L}^L f(x) d x \\
a_n=\frac{1}{L} \int_{-L}^L f(x) \cos \frac{n \pi x}{L} d x \\
b_n=\frac{1}{L} \int_{-L}^L f(x) \operatorname{sen} \frac{n \pi x}{L} d x
\end{gathered}
$$

**Exemplo 1:** Usando Python para encontrar a série de Fourier da função (Onda Retangular Periódica)

$$
f(x)=\left\{\begin{array}{llc}
0 & \text { se }-2<x<-1 \\3  & \text { se }-1<x<1 \\
0 & \text { se } 1<x<2
\end{array} \quad p=2 L=4, \quad L=2 .\right.
$$



```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x):
    if abs(x)>2: 
        return f(abs(x)-4)
    if (x>0)&(x<1) or (x<0)&(x>-1): 
        return 3
    else:
        return 0
    
fig, ax = plt.subplots(figsize=(8,2))
xi = np.linspace(-6,6,1000)
yi = [f(x) for x in xi]
plt.plot(xi,yi)
plt.grid()
```

Aplicando as fórmulas de Euler obtemos $a_0=3 / 2$

$$
a_n=\frac{1}{2} \int_{-2}^2 f(x) \cos \frac{n \pi x}{2} d x=\frac{1}{2} \int_{-1}^13 \cos \frac{n \pi x}{2} d x=\frac{2 k}{n \pi} \operatorname{sen} \frac{n \pi}{2}
$$

Portanto, $a_n=0$ e  se $n$ for par e

$$
a_n=2 k / n \pi \quad \text { se } \quad n=1,5,9, \cdots, \quad a_n=-2 k / n \pi \quad \text { se } \quad n=3,7,11, \cdots .
$$

Tem-se também que $b_n=0$ para $n=1,2, \cdots$ Logo, a série de Fourier é

$$
f(x)=\frac{3}{2}+\frac{6}{\pi}\left(\cos \frac{\pi}{2} x-\frac{1}{3} \cos \frac{3 \pi}{2} x+\frac{1}{5} \cos \frac{5 \pi}{2} x-+\cdots\right) .
$$

A série de Fourier de uma função par de período $2 L$ é uma **série de Fourier de cossenos**

$$
f(x)=a_0+\sum_{n=1}^{\infty} a_n \cos \frac{n \pi}{L} x
$$

com os coeficientes
$$
a_0=\frac{1}{L} \int_0^L f(x) d x, \quad a_n=\frac{2}{L} \int_0^L f(x) \cos \frac{n \pi x}{L} d x, \quad n=1,2, \cdots
$$

A série de Fourier de uma função ímpar de período $2 L$ é uma **série de Fourier de senos**

$$
f(x)=\sum_{n=1}^{\infty} b_n \operatorname{sen} \frac{n \pi}{L} x
$$

com os coeficientes

$$
b_n=\frac{2}{L} \int_0^L f(x) \operatorname{sen} \frac{n \pi x}{L} d x
$$

#### Expansões de Meia-escala

Uma função par pode ser representada por uma série formada apenas por termos em cosseno, enquanto uma função ímpar pode ser representada por uma série formada apenas por termos em seno. Assim, para expandir uma função $f(x)$ em série de senos ou de cossenos, pode ser útil estendê-la a uma função periódica de período $2L$, escolhendo uma extensão par ou ímpar, conforme for mais conveniente. Essa construção recebe o nome de **expansão de meia-escala**, que corresponde a uma extensão periódica par ou ímpar, como ilustrado na figura a seguir.

### Equação da onda

Dada a EDP

$$
\frac{\partial^2 u}{\partial t^2}=c^2 \frac{\partial^2 u}{\partial x^2}
$$

sob as condições 

$$
u(0, t)=0, \quad u(L, t)=0 \quad \text { para todo } t \quad \text { e}
$$
$$
u(x, 0)=f(x), \quad u_t(x, 0)=g(x) \quad(0 \leqq x \leqq L)
$$

Sabe-se que a solução analítica é dada por 

$$
u(x, t)=\sum_{n=1}^{\infty} u_n(x, t)=\sum_{n=1}^{\infty}\left(B_n \cos \lambda_n t+B_n * \operatorname{sen} \lambda_n t\right) \operatorname{sen} \frac{n \pi}{L} x
$$

com 

$$B_n=\frac{2}{L} \int_0^L f(x) \operatorname{sen} \frac{n \pi x}{L} d x$$

Sabe-se que a solução analítica é dada por 

$$
u(x, t)=\sum_{n=1}^{\infty} u_n(x, t)=\sum_{n=1}^{\infty}\left(B_n \cos \lambda_n t+B_n * \operatorname{sen} \lambda_n t\right) \operatorname{sen} \frac{n \pi}{L} x
$$

com $\lambda_n =\frac{c n \pi}{L}$ e 

$$B_n=\frac{2}{L} \int_0^L f(x) \operatorname{sen} \frac{n \pi x}{L} d x \quad \text{e} \quad B_n^*=\frac{2}{c n \pi} \int_0^L g(x) \operatorname{sen} \frac{n \pi x}{L} d x$$

## Diferenças finitas para EDOs hiperbólicas
Fonte: (https://integrada.minhabiblioteca.com.br/reader/books/9788522123414/pageid/847)


Agora, consideramos a resolução numérica da equação da onda, um exemplo de uma equação diferencial parcial hiperbólica. A equação da onda é dada pela equação diferencial

$$
\frac{\partial^2 u}{\partial t^2}(x, t)-\alpha^2 \frac{\partial^2 u}{\partial x^2}(x, t)=0, \quad 0<x<l, \quad t>0
$$

sujeita às condições

$$
u(0, t)=u(l, t)=0, \quad \text { para } t>0 {, }
$$

$$
u(x, 0)=f(x) \quad \text { e } \quad \frac{\partial u}{\partial t}(x, 0)=g(x), \quad \text { para } 0 \leq x \leq l \text {, }
$$

em que a é uma constante que depende das condições físicas do problema.

Em qualquer ponto interior da malha, $(x_i, t_j)$, a equação da onda torna-se

$$
\frac{\partial^2 u}{\partial t^2}\left(x_i, t_j\right)-\alpha^2 \frac{\partial^2 u}{\partial x^2}\left(x_i, t_j\right)=0 {.}
$$

que leva aà equação de diferença

$$
\frac{w_{i, j+1}-2 w_{i, j}+w_{i, j-1}}{k^2}-\alpha^2 \frac{w_{i+1, j}-2 w_{i, j}+w_{i-1, j}}{h^2}=0 .
$$

Fazendo $\lambda=\alpha k / h$ podemos escrever a equação de diferença como

$$
w_{i, j+1}-2 w_{i, j}+w_{i, j-1}-\lambda^2 w_{i+1, j}+2 \lambda^2 w_{i, j}-\lambda^2 w_{i-1, j}=0
$$

isolando $w_{i,j+1}$ obtemos

$$
w_{i, j+1}=2\left(1-\lambda^2\right) w_{i, j}+\lambda^2\left(w_{i+1, j}+w_{i-1, j}\right)-w_{i, j-1} .
$$

Os valores para $j=1$, que são necessários na equação de diferença, para calcular $w_{i,2}$,devem ser obtidos da condição de velocidade inicial 

$$
\frac{\partial u}{\partial t}(x, 0)=g(x), \quad 0 \leq x \leq l .
$$

A qual é usada para obter a equação de diferença

$$
w_{i, 1}=\left(1-\lambda^2\right) f\left(x_i\right)+\frac{\lambda^2}{2} f\left(x_{i+1}\right)+\frac{\lambda^2}{2} f\left(x_{i-1}\right)+k g\left(x_i\right),
$$

que pode ser usada para encontrar $w_{i,1}$, para cada $i=1,2,...,m-1$.

Ver Burden e Faires (2016).

```python
l = 1.
T = 1.
alpha = 3.87
m = 10
N = 200
```

```python
h = l/m
k = T/N
lamb = k*alpha/h
lamb
```

Esses métodos podem ser aplicados à **equação de Burgers**

$$
\frac{\partial u}{\partial t} + u \frac{\partial u}{\partial x} = 0
$$

pois ela é uma equação hiperbólica não trivial e não linear. Considerando a condição inicial

$$
u(x, 0) = u_0(x) = 
\begin{cases}
1, & x_0 \leq 0 \\
1 - x, & 0 \leq x_0 \leq 1 \\
0, & x_0 \geq 1
\end{cases}
$$

a linha característica que parte do ponto $(x_0, 0)$ é dada por

$$
x(t) = x_0 + t\,u_0(x_0) = 
\begin{cases}
x_0 + t, & x_0 \leq 0 \\
x_0 + t(1 - x_0), & 0 \leq x_0 \leq 1 \\
x_0, & x_0 \geq 1
\end{cases}
$$


**Referências**

```python
def f(x):
    return np.sin(np.pi * x) 

def g(x):
    return 0  

def finite_difference(f, l, t, m, N, alpha, a, b):
    h = l / m
    k = t / N
    lamb = alpha * k / h

    w = np.zeros((N+1, m+1))

    for j in range(1, N+1):
        w[j, 0] = a
        w[j, m] = b 

    for i in range(1, m):
        w[0, i] = f(i * h)  
        w[1, i] = (1 - lamb**2) * f(i * h) + ((lamb**2) / 2) * (f((i+1) * h) + f((i-1) * h)) + k * g(i * h)

    for j in range(1, N):
        for i in range(1, m):
            w[j+1, i] = 2 * (1 - lamb**2) * w[j, i] + (lamb**2) * (w[j, i+1] + w[j, i-1]) - w[j-1, i]

    return w

l, T = 1, 1
m, N = 10, 20
alpha = 2
a, b = 0, 0

w = finite_difference(f, l, T, m, N, alpha, a, b)

x = np.linspace(0, l, m+1) 
t = np.linspace(0, T, N+1) 

for j in range(0, N+1, 5): 
    plt.plot(x, w[j, :], label=f't = {t[j]:.2f}s')

plt.xlabel('Distância (m)')
plt.ylabel('Deslocamento (u)')
plt.title('Diferenças Finitas Equação da Onda')
plt.legend()
plt.grid(True)
plt.show()
```

prompt Gemini: "me mostra um script python para resolver um problema da equação da onda por diferenças finitas"

```python
def resolver_equacao_da_onda(L, T, c, nx, nt):
    """
    Resolve a equação da onda 1D usando o método de diferenças finitas.

    Parâmetros:
    L (float): Comprimento do domínio espacial.
    T (float): Tempo total da simulação.
    c (float): Velocidade de propagação da onda.
    nx (int): Número de pontos na malha espacial.
    nt (int): Número de passos de tempo.

    Retorna:
    tuple: (x, t, u), onde x é o vetor de posições, t é o vetor de tempos,
           e u é a matriz da solução (u[t, x]).
    """
    # Discretização
    dx = L / (nx - 1)
    dt = T / (nt - 1)

    # Coeficiente de estabilidade de Courant-Friedrichs-Lewy (CFL)
    # Deve ser <= 1 para garantir estabilidade numérica.
    CFL = c * dt / dx

    if CFL > 1:
        print(f"Aviso: O número de Courant é {CFL:.2f}, maior que 1.")
        print("A simulação pode se tornar instável. Considere aumentar nx ou diminuir nt.")
        # É uma boa prática não parar a execução, mas alertar o usuário.

    # Inicialização das malhas
    x = np.linspace(0, L, nx)
    t = np.linspace(0, T, nt)
    u = np.zeros((nt, nx))

    # Condição inicial: uma onda senoidal (exemplo)
    def u_0(x):
        return np.sin(2 * np.pi * x / L)

    u[0, :] = u_0(x)

    # Condição inicial de velocidade (velocidade inicial zero)
    def v_0(x):
        return np.sin(4 * np.pi * x / L)

    # Passo de tempo 1 (para inicializar a iteração principal)
    # Usa a aproximação de diferença central para a derivada de tempo
    # u[1, i] = u[0, i] + dt * v_0(x[i])
    for i in range(1, nx - 1):
        u[1, i] = u[0, i] + dt * v_0(x[i]) + 0.5 * CFL**2 * (u[0, i+1] - 2 * u[0, i] + u[0, i-1])

    # Condições de contorno (exemplo: extremidades fixas)
    u[:, 0] = 0
    u[:, nx - 1] = 0

    # Loop principal de diferenças finitas
    # Usa a fórmula de diferenças finitas para a equação da onda
    # u[t+1, x] = 2 * u[t, x] - u[t-1, x] + CFL^2 * (u[t, x+1] - 2 * u[t, x] + u[t, x-1])
    for n in range(1, nt - 1):
        for i in range(1, nx - 1):
            u[n + 1, i] = 2 * u[n, i] - u[n - 1, i] + CFL**2 * (u[n, i + 1] - 2 * u[n, i] + u[n, i - 1])

    return x, t, u

# --- Parâmetros da Simulação ---
L = 1.0  # Comprimento da corda
T = 2.0  # Tempo total
c = 1.0  # Velocidade da onda
nx = 101 # Número de pontos na malha espacial
nt = 201 # Número de passos de tempo

# --- Execução da Simulação ---
x, t, u = resolver_equacao_da_onda(L, T, c, nx, nt)

# --- Visualização dos Resultados ---

# Plot da solução em um instante de tempo específico
plt.figure(figsize=(10, 6))
plt.plot(x, u[0, :], label="t = 0.0")
plt.plot(x, u[nt//4, :], label=f"t = {t[nt//4]:.2f}")
plt.plot(x, u[nt//5, :], label=f"t = {t[nt//5]:.2f}")
plt.plot(x, u[nt//10, :], label=f"t = {t[nt//10]:.2f}")
plt.plot(x, u[nt//2, :], label=f"t = {t[nt//2]:.2f}")
plt.title("Solução da Equação da Onda em Diferentes Tempos")
plt.xlabel("Posição (x)")
plt.ylabel("Deslocamento (u)")
plt.grid(True)
plt.legend()
plt.show()
```
