## 2.1 Localização de raízes

Em ciências e engenharias frequentemente a solução de um problema real é obtida resolvendo uma equação em uma variável real, a qual sempre pode ser escrita na forma 

$$f(x)=0$$

Dada uma função $f:R \rightarrow R$ dizemos que $\overline{x} \in R$, tal que $f(\overline{x})=0$ é um _zero_ (ou _raiz_) da função $f$. Graficamente, uma raiz é o ponto onde a função $f(x)$ cruza ou toca o eixo $x$. Uma equação $f(x)=0$ pode não ter solução (raiz) real, ter uma única raiz ou várias.


![](https://raw.githubusercontent.com/tiagoburiol/metodos_numericos_em_python/refs/heads/main/2_ZEROS_DE_FUNCOES/imagens/raizes.jpg){width=600px}


Quando a equação é simples, o valor de $x$ pode ser determinado analiticamente. Esse é o caso quando se escreve $x$ explicitamente após a aplicação de operações matemáticas, ou quando uma fórmula conhecida (como a fórmula de Bháskara para resolução de equações quadráticas) pode ser usada para determinar o valor exato de $x$. Em muitas situações, no entanto, é impossível determinar analiticamente a raiz de uma equação.

Por exemplo, é relativamente fácil obter analiticamente as raízes das equações 

&nbsp;&nbsp;&nbsp;&nbsp; a) $2.1x-47.3 = 0$

&nbsp;&nbsp;&nbsp;&nbsp; b) $x^2+2x-3 = 0$

&nbsp;&nbsp;&nbsp;&nbsp; c) $senx-x = 0$

Já para as equações a seguir é preferível utilizar-se de métodos numéricos

&nbsp;&nbsp;&nbsp;&nbsp; a) $xe^x-2 = 0$

&nbsp;&nbsp;&nbsp;&nbsp; b) $cos3\theta-\operatorname{sen}\theta =0$

&nbsp;&nbsp;&nbsp;&nbsp; c) $2.1^x+5.7x=0$

**Exemplo 2.1.1:**
Em termodinâmica, sob determinadas condições, a relação entre o calor $Q$ fornecido a um gás e sua variação de temperatura $T_f - T_i$ é dada por 

$$ Q = nR \left[ A (T_f-T_i)+ \frac{B}{2}(T_f^2-T_i^2) + \frac{C}{3}(T_f^3-T_i^3) \right].$$

Para o gás metano tem-se $R=8,314 J/mol.K$, $A=1,702$, $B=9,081 \times 10^{-3}K^{-1}$, $C=-2,164 \times 10^{-6} K^{-2}$. 

Considere o problema de calcular a temperatura final $T_f$ se $20 kJ$ de energia forem absorvidos pelo gás, supondo estar contido em uma câmara com $n=2 mol$ de metano a temperatura $T_i = 300 K$. 

**Exemplo 2.1.2:**
A área do segmento $A_S$ de um círculo de raio $r$ (ver figura abaixo) é dada por $A_S=\frac{1}{2}r^2(\theta-\operatorname{sen}\theta).$

Para se determinar o ângulo $\theta$ se $A_S$ e $r$ são conhecidos, a equação deve ser resolvida para $\theta$. Não é uma tarefa simples escrever $\theta$ explicitamente em termos de $A_S$ e $r$, portanto, pode não ser possível resolver essa equação analiticamente.

**Exemplo 2.1.3:** Seja uma esfera de madeira de raio $r$ colocada a flutuar sobre a água conforme figura. Qual é a altura $d$ da esfera que ficará submersa? 


![](https://raw.githubusercontent.com/tiagoburiol/metodos_numericos_em_python/refs/heads/main/2_ZEROS_DE_FUNCOES/imagens/esfera.png){width=260px}

Para responder essa pergunta, observemos que, pelo *Princípio de Arquimedes* o peso da esfera deve ser equilibrado com o peso do volume de água deslocada, ou seja, $\mu_e V_e g = \mu_a V_a g$ onde $g$ é a aceleração da gravidade, $\mu_e$ é a densidade relativa da madeira e $\mu_a$ é a densidade relativa da água, $V_e$ e $V_a$ são os volumes. Essa equação nos leva a 

$$ \mu_e \frac{4 \pi r^3}{3} = \mu_a \int_0^d \pi \left[r^2-(x-r)^2) \right]dx $$

$$ \mu_e \frac{4 \pi r^3}{3} = \mu_a \frac{\pi d^2 (3r-d)}{3}$$


substituindo os valores $r=10 cm$,  $\mu_e =0,638 g/cm^3$ e $\mu_a=1 g/cm^3$, com $\pi \approx 3.1415$, podemos chegar à seguinte equação aproximada

$$ 1.047d^3-31.415d^2+2672.369=0$$

Basta resolver essa equação para obter a resposta procurada. 

No entanto, encontrar a solução analítica para esse problema não é tão simples, mas com uma rotina numérica simples podemos encontrar uma boa aproximação. 

Observe que poderíamos ter multiplicado ambos os lados da equação por $\frac{3}{\pi}$, o que nos daria  

$$d^3-30d^2+2552=0$$

que é uma equação mais precisa, uma vez que não tivemos que usar aproximações, no entanto esse tipo de simplificação requer manipular o número $\pi$ como um símbolo. Veremos que frequentemente soluções numéricas geram soluções aproximadas 

**Exemplo 2.1.4:** Podemos utilizar uma abordagem numérica para obter soluções aproximadas de equações, vejamos por exemplo a equação $f(d) = 1.047d^3-31.415d^2+2672.369$ do exemplo descrito anteriormente. Dos dados do problema, podemos inferir que a solução é um número entre 0 e 20, então, calculamos valores da função nesse intervalo e tentar encontrar uma aproximação. Buscamos o valor de $d$ que anula a função $f$, então, observando os resultados na tabela abaixo, podemos concluir que a solução é um número entre 11 e 12 cm pois é o intervalo onde a função muda de sinal.  

```python
for d in range(0,20):
    y = 1.047*d**3-31.415*d**2+2672.369
    print(f'f({d}) = {round(y, 3)}')
```

Seguindo o mesmo procedimento mas agora incrementando o valor de $d$ em 0.1 e calculando $f$ para valores de $d$ entre 11 e 12 observamos que a solução está entre $11.8$ e $11.9$. 

```python
for inc in range(0,11):
    d = 11+inc/10
    y = 1.047*d**3-31.415*d**2+2672.369
    print('f(',d,')=', round(y,3))
```

Por fim, podemos tomar como solução aproximada o ponto médio entre $11.8$ e $11.9$, ou seja, $d=11.85$. A precisão é de $0.05$.

**Exemplo 2.1.5:** (Usando a biblioteca SciPy) Encontrando a raiz de $A_S=\frac{1}{2}r^2(\theta-\operatorname{sen}\theta)$ com $r=1$ e $A_s = 3.5$.

```python
from scipy.optimize import fsolve
import numpy as np

As = lambda x: 0.5*(x-np.sin(x))-3.5

sol = fsolve(As, 1)
print (sol)
```

Esses foram apenas alguns exemplos de como os métodos numéricos podem nos auxiliar na obtenção de soluções aproximadas quando não é fácil, ou mesmo possível, encontrar soluções analíticas. No entanto, sua aplicação vai muito além desses casos. Os métodos numéricos são amplamente utilizados na resolução de integrais definidas, sistemas lineares de grande porte, equações diferenciais, problemas de otimização e na visualização e análise de grandes conjuntos de dados, entre muitas outras aplicações.

Nos casos em que uma solução exata $\overline{x}$ da equação $f(x)=0$ não puder ser encontrada por processos analíticos, uma solução aproximada poderá ser obtida por meio de métodos numéricos. Em geral, esses métodos produzem uma sequência de aproximações, obtidas por um processo iterativo que é encerrado quando a precisão desejada é atingida.

De forma geral, a determinação numérica de uma raiz pode ser dividida em duas etapas:

1. **Isolamento da raiz:** consiste em encontrar um intervalo $[a,b]$ que contenha uma única raiz da equação $f(x)=0$;

2. **Refinamento da aproximação:** consiste em reduzir sucessivamente o intervalo que contém a raiz até que a precisão desejada seja alcançada.

Para isolar as raízes de uma função, podemos utilizar métodos gráficos ou avaliar a função em diferentes pontos do domínio, identificando intervalos nos quais ocorre mudança de sinal.


### Isolamento da raiz: Localização por métodos gráficos

Outra abordagem seria atribuir valores arbitrários para a variável $x$ e encontrar um par $a$ e $b$ tal que $f(a)\cdot f(b)<0$.

### Localização de raízes de polinômios

Para funções polinomiais, além dos métodos gráficos, existem teoremas da Álgebra que permitem investigar a existência e a localização de suas raízes. Alguns desses resultados serão apresentados a seguir.

Desejamos localizar as raízes $\overline{x}$ tais que $P(\overline{x})=0$, em que $P(x)$ é um polinômio de grau $n \in \mathbb{N}$, com coeficientes reais ou complexos, da forma

$$
P(x)=a_0+a_1x+a_2x^2+\cdots+a_nx^n,
$$

em que $a_n \neq 0$.

As raízes de um polinômio satisfazem as seguintes propriedades:

1. Um polinômio de grau $n$ possui exatamente $n$ raízes reais ou complexas, contando suas multiplicidades. Assim, algumas raízes podem coincidir.

2. Se os coeficientes do polinômio forem reais e $n$ for ímpar, então existe pelo menos uma raiz real.

3. Se os coeficientes do polinômio forem reais e existirem raízes complexas, elas ocorrerão aos pares conjugados. Ou seja, se $a+bi$ é uma raiz, então $a-bi$ também é raiz, em que $i=\sqrt{-1}$.

O valor de uma função polinomial pode ser calculado de forma mais eficiente quando o polinômio é reescrito na chamada **forma estruturada**, também conhecida como forma de **parênteses aninhados** (ou forma de Horner). Nessa representação, reduz-se significativamente o número de operações aritméticas necessárias para avaliar o polinômio.

Por exemplo, considere o polinômio de terceiro grau

$$
f_3(x)=a_3x^3+a_2x^2+a_1x+a_0.
$$

Se as potências de $x$ forem calculadas por multiplicações sucessivas, sua avaliação exige seis multiplicações e três adições. De modo geral, um polinômio de grau $n$ requer  $\frac{n(n+1)}{2}+n
$ multiplicações e $n$ adições.

Entretanto, se o polinômio for reescrito na forma

$$
f_3(x)=\left(\left(a_3x+a_2\right)x+a_1\right)x+a_0,
$$

serão necessárias apenas três multiplicações e três adições. Essa forma reduz o custo computacional e é conhecida como **método de Horner**.

**Exemplo 2.1.8:** Considere o polinômio $f(x)=0.8x^5 - 1.3x^4 +5x^3 +17x^2 + 1.9x +10$. Vamos calcular $f(0.5)$ usando um laço simples em que cada passo efetuamos uma multiplicação e uma soma.

```python
a = [0.8, -1.3,5, 17,1.9,10] #coeficientes (do maior para o menor grau)
x = 0.5 
```

A variável $p$ armazenará o valor parcial do polinômio obtido pelo método de Horner.

```python
p = 0
for n in range(6):
    p = p*x+a[n]
print(p)
```

Uma vantagem do método de Horner é que ele também permite calcular o valor da derivada com um custo computacional muito pequeno. Basta acrescentar uma variável para armazenar o valor parcial da derivada. Vamos fazer isso incluindo a linha `df = df*x+p`no laço acima.

```python
p = 0
dp = 0

for n in range(6):
    dp = dp*x+p
    p = p*x+a[n]
    
print(f"f({x})  = {p}")
print(f"f'({x}) = {dp}")
```

Podemos confirmar os valores obtidos calculando diretamente o polinômio e sua derivada.

```python
p = 0.8*x**5-1.3*x**4+5*x**3+17*x**2+1.9*x+10
df = 5*0.8*x**4-1.3*4*x**3+3*5*x**2+2*17*x+1.9
print (p, df)
```

Os valores coincidem com aqueles obtidos pelo método de Horner, confirmando a correção da implementação.

**Observação:** Seja $P(x)$ um polinômio de grau $n \geq 1$. Dizemos que $\bar{x}$ é uma raiz de multiplicidade " $m$ " da equação $P(x)=0$ se:

$$
P(\bar{x})=P^{(1)}(\bar{x})=\ldots=P^{(m-1)}(\bar{x})=0 \quad \mathrm{e} \quad P^{(m)}(\bar{x}) \neq 0
$$

em que $P^m(\bar{x})$ é a $m$-ésima derivada de $P(x)$ calculada no ponto $\bar{x}$.

Por exemplo, considere o polinômio $P(x)=x^3$. A raiz $\bar{x}=0$ é uma raiz de multiplicidade três da equação $P(x)=0$, pois:

$$
P(x)=P^{(1)}(\bar{x})=0, \quad P^{(2)}(\bar{x})=0 \quad \text { e } \quad P^{(3)}(\bar{x})=6 \neq 0
$$

Entendemos, neste caso, que a equação $P(x)=x^3=0$ possui três raízes idênticas, ou uma raiz de multiplicidade 3 .

A seguir apresentamos alguns teoremas que podem ser utilizados para investigar a localização das raízes de polinômios.

#### Teorema: Localização de raízes no círculo

Sejam $P(x)=a_0 x^n + a_1 x^{n-1}+\cdots+a_{n-1}x+a_n$ e $A=\max\{\left|a_1 \right|,\ldots, \left|a_n \right|\}$. Então, o módulo de todas as raízes $x_k$ , $k = 1,\ldots, n$ satisfaz a inequação:

$$ |x_k|< 1 +\frac{A}{\left|a_0 \right|} = R$$

Para a demonstração, ver Arenales e Darezzo, 2016.

Como consequência desse teorema, tem-se o seguinte **corolário**: 

Sejam $a_n\neq 0$ e $B=\max\{\left|a_0 \right|,\ldots,\left|a_{n-1}\right|\}$, em que $a_k$,$k=0,\ldots,n$ são os coeficientes de $P(x)$. Então, o módulo de todas as raízes $x_k$, $k = 1,\ldots, n$ satisfaz a inequação:

$$ \left|x_k \right| > \frac{1}{1+\frac{B}{\left|a_n \right|}}=r$$

(Para a demonstração, ver Arenales e Darezzo, 2015, p 123)

![](https://raw.githubusercontent.com/tiagoburiol/metodos_numericos_em_python/refs/heads/main/2_ZEROS_DE_FUNCOES/imagens/anel.png){width=250px}


**Exemplo 2.1.9:** (Fonte: [Arenales e Darezzo, 2016](https://integrada.minhabiblioteca.com.br/reader/books/9788522112821/pageid/136)) As raízes de $P(x)=x^3+2x^2-x-2=0$ estão localizadas em $-3<x_k< -1/2$ ou $1/2< x_k <3$, pois:

$$ |x_k|<1-\frac{2}{|1|} = 3 = R \quad \text{e} \quad |x_k|> \frac{1}{1+\frac{2}{|-2|}} = 1/2 = r$$
 
De fato, as raízes de $P(x) = 0$ são dadas por $x = 1$, $x = −1$ e $x = −2$.

Assim, no plano complexo as raízes de $P(x) = 0$ estão localizadas no anel $r < |x| < R$ como ilustra a figura abaixo.


### Teorema: regra de sinais de Descartes

Dada uma equação polinomial com coeficientes reais, o número de raízes reais positivas $p$ dessa equação não excede o número $v$ de variações de sinais dos coeficientes. Ainda mais, $(v − p)$ é inteiro par, não negativo.

Para determinar o número de raízes negativas, basta calcular o número de variações de sinais no polinômio $P(−x)$.

### Teorema: Budan-Fourier

Se os números $a$ e $b(a<b)$ não são raízes da equação $P(x)=0$, então o número $N(a, b)$ de raízes reais da equação $P(x)=0$ localizadas entre $a$ e $b$ é dado por $N(a, b)=\Delta \bar{N}-2 k \quad(k$ natural $)$, sendo $\Delta \bar{N}=\bar{N}(a)-\bar{N}(b)$ em que $\bar{N}(x)$ é igual ao número de variações de sinais na sequência de sucessivas derivadas:

$$
P(x), P^{(1)}(x), \ldots, P^{(n-1)}, 1
$$

**Exemplo 2.1.10:** (Fonte: [Arenales e Darezzo, 2016](https://integrada.minhabiblioteca.com.br/reader/books/9788522112821/pageid/137)) Considere o polinômio $P(x)=x^3+2 x^2-x-2$. Aplicando o teorema de Budan-Fourier, vemos que os valores 3 e -3 não são raízes de $P(x)=0$, pois $P(3)=40$ e $P(-3)=-8$. Desta forma, é possível calcular o número de raízes localizadas em $[-3,3]$ :

Seja $P^{(i)}(x) \rightarrow i$-ésima derivada de $P(x)$. Assim, temos:

$$
\begin{array}{rrrr}
P(3)=40 & P^{(1)}(3)=38 & P^{(2)}(3)=22 & P^{(3)}(3)=6\\
P(-3)=-8 & P^{(1)}(-3)=14 & P^{(2)}(-3)=-14 & P^{(3)}(-3)=6
\end{array}
$$
Logo,

$$
\begin{gathered}
\bar{N}(3)=0 \quad \text { e } \quad \bar{N}(-3)= 3\\
\Delta \bar{N}=\bar{N}(-3)-\bar{N}(3)=3-0 = 3\\
N(-3,3)=3-2 k, \quad k=0,1
\end{gathered}
$$


Portanto, temos uma ou três raízes no intervalo. Pela regra de sinais de Descartes, observamos que a equação possui uma raiz positiva e, portanto, a equação pode ter duas raízes no intervalo $(-3,0)$ ou não ter nenhuma raiz real negativa.
